```python
import mlflow
from mlflow import pyfunc
import mlflow
mlflow.set_tracking_uri("http://localhost:5000")  # Make sure this is the same everywhere

# Example: Load and register the model (if not registered already)
model = pyfunc.load_model("models:/FakeNewsModel/Production")
model_name = "FakeNewsModel"

client = mlflow.tracking.MlflowClient()

# Check if the model exists, and if not, register it
try:
    client.get_registered_model(model_name)
except mlflow.exceptions.MlflowException:
    print(f"Registering new model: {model_name}")
    client.create_registered_model(model_name)

# Log the model to MLflow
client.create_model_version(model_name, "path_to_model", "1")
```
